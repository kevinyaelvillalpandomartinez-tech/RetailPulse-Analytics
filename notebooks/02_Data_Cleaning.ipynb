## Identificación de variables categóricas

Antes de convertir las columnas al tipo `category`, se evaluó el número de valores únicos de cada variable.

Las columnas seleccionadas presentan un número reducido de categorías repetidas, por lo que su conversión permitirá optimizar el consumo de memoria sin alterar la información almacenada.

Las columnas `Customer Name`, `Product Name`, `Customer ID` y `Product ID` se mantendrán como texto debido a su alta cardinalidad o porque funcionan como identificadores.

```python
categorial_colums = ["Ship Mode", "Segment","Country","City","State","Region","Category","Sub-Category"]
for col in categorial_colums:
    df_clean[col] = df_clean[col].astype("category")
```

## Optimización del uso de memoria

Se evaluó la cardinalidad de las variables categóricas y se identificaron ocho columnas susceptibles de convertirse al tipo `category`.

Las columnas convertidas fueron:

- Ship Mode
- Segment
- Country
- City
- State
- Region
- Category
- Sub-Category

Antes de la optimización, el conjunto de datos consumía aproximadamente **9.2 MB** de memoria.

Después de convertir estas variables al tipo `category`, el consumo se redujo a **4.3 MB**, lo que representa una disminución aproximada del **53 %**.

Esta optimización mejora el rendimiento del procesamiento de datos sin modificar la información contenida en el dataset.
